# Rename Directories
This notebook renames the project folders from client/server to frontendd/bckend.

```python
from pathlib import Path
root = Path(r'c:\Users\Kavyasri\OneDrive\Desktop\nxtwave project')
mapping = {'client': 'frontendd', 'server': 'bckend'}
for old_name, new_name in mapping.items():
    old_path = root / old_name
    new_path = root / new_name
    if old_path.exists():
        if new_path.exists():
            print(f'{new_name} already exists, skipping rename')
        else:
            old_path.rename(new_path)
            print(f'Renamed {old_name} to {new_name}')
    else:
        print(f'{old_name} not found')
print('---')
print([p.name for p in root.iterdir() if p.is_dir()])
```
