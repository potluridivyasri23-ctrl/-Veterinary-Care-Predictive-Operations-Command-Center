```python
# Rename Files in Directory
This notebook renames files in a directory using Python and confirms the results.
</VSCode.Cell>

<VSCode.Cell id="2" language="python">
import os
from pathlib import Path

# Set the directory path for renaming
target_dir = Path(r"c:\Users\Kavyasri\OneDrive\Desktop\nxtwave project")
</VSCode.Cell>

<VSCode.Cell id="3" language="markdown">
## List Files in Directory
Display the current filenames in the target directory.
</VSCode.Cell>

<VSCode.Cell id="4" language="python">
files = sorted([p.name for p in target_dir.iterdir() if p.is_file()])
print("Current files:")
for f in files:
    print(f)
</VSCode.Cell>

<VSCode.Cell id="5" language="markdown">
## Define Rename Mapping Logic
Create a function that maps old filenames to new filenames.
</VSCode.Cell>

<VSCode.Cell id="6" language="python">
def get_new_name(old_name: str) -> str:
    mapping = {
        "client": "frontend",
        "server": "backend",
        "rename-folders.bat": "rename-folders-helper.bat",
        "rename-folders.js": "rename-folders-helper.js",
    }
    return mapping.get(old_name, old_name)

# Example mapping check
print(get_new_name("client"))
print(get_new_name("README.md"))
</VSCode.Cell>

<VSCode.Cell id="7" language="markdown">
## Rename Files Safely
Rename files using pathlib.Path.rename and skip collisions.
</VSCode.Cell>

<VSCode.Cell id="8" language="python">
for file_path in sorted(target_dir.iterdir()):
    if file_path.is_file():
        new_name = get_new_name(file_path.name)
        if new_name != file_path.name:
            new_path = target_dir / new_name
            if new_path.exists():
                print(f"Skipping rename; target already exists: {new_path}")
            else:
                print(f"Renaming {file_path.name} -> {new_name}")
                file_path.rename(new_path)
</VSCode.Cell>

<VSCode.Cell id="9" language="markdown">
## Verify Renamed Files
List the files again after renaming to confirm the updates.
</VSCode.Cell>

<VSCode.Cell id="10" language="python">
updated_files = sorted([p.name for p in target_dir.iterdir() if p.is_file()])
print("Updated files:")
for f in updated_files:
    print(f)
```
